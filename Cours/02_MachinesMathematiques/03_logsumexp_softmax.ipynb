$$
\frac1\beta\log\left(\sum_j e^{\beta x_j}\right)

## Summary

Softmax is the gradient of a smooth convex approximation of the maximum. Its Hessian is a weighted covariance matrix: it is positive semidefinite and annihilates the constant direction. Finally, subtracting the maximum before computing exponentials leaves the function unchanged but makes evaluation much more robust.
