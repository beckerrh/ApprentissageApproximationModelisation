## Bilan

La SVD donne explicitement le meilleur sous-espace de reconstruction linéaire pour l'erreur quadratique. Un autoencodeur linéaire apprend une factorisation de cette reconstruction par une optimisation non convexe ; il peut retrouver le même sous-espace sans que ses facteurs soient uniques.

L'intérêt véritable des autoencodeurs apparaîtra lorsque l'encodeur ou le décodeur seront non linéaires : l'ensemble reconstruit ne sera alors plus limité à un sous-espace vectoriel.
