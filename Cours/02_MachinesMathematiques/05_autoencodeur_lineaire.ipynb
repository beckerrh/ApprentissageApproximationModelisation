## Summary

The SVD explicitly gives the best linear reconstruction subspace for squared error. A linear autoencoder learns a factorisation of this reconstruction through non-convex optimisation; it can recover the same subspace without its factors being unique.

The real benefit of autoencoders will emerge when the encoder or decoder is non-linear: the reconstruction set will then no longer be restricted to a vector subspace.
