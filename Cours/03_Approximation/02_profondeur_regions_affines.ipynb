## 1. A $P^1$ machine written using ReLU

A piecewise affine function on the real line has a representation
$$
v(x)=\alpha+\beta x+\sum_{i=1}^N c_i(x-t_i)_+.
$$

The nodes $t_i$ are the points where the slope may change; $c_i$ is precisely the slope jump at $t_i$. The following function is the minimal version of the machine constructed in notebook 03.

```python
def machine_p1_relu(x, alpha, beta, noeuds, sauts_pente):
    activations = jax.nn.relu(x[..., None] - noeuds)
    return alpha + beta * x + activations @ sauts_pente
```

## Summary

The tent function is a $P^1$ machine, hence a small ReLU network. Its self-composition remains easy to evaluate, but its description as a piecewise affine function quickly becomes complex.

To represent $\tau^{\circ L}$, choose constant width $3$ and depth $L$, or return to a single hidden layer with width of order $2^L$. In this example, depth therefore replaces exponential growth in width.
