## Bilan

Un plongement est une factorisation de la dépendance aux tokens par un espace intermédiaire. Lorsque $d\ll V$, cette structure réduit le nombre de paramètres et partage de l'information. Elle remplace toutefois une variable matricielle par deux facteurs et introduit une paramétrisation non convexe et non unique.
