## Bilan

Le masque causal impose une structure triangulaire : la sortie en position $i$ ne dépend que des valeurs d'indices $j\leq i$. Le softmax transforme ensuite chaque ligne de scores en coefficients positifs de somme un, et la multiplication par $V$ forme la combinaison correspondante.
