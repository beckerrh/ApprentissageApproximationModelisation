## Exercice 1.1 — Produit de Hadamard

Pour $x,y\in\mathbb R^n$, le produit de Hadamard est

$$
x\odot y=(x_i y_i)_{1\leq i\leq n}.
$$

Pour $x,y,z\in\mathbb R^n$, montrer que

$$
x\odot y=\operatorname{Diag}(x)y=\operatorname{Diag}(y)x,
$$

$$
\operatorname{Diag}(x\odot y)
=\operatorname{Diag}(x)\operatorname{Diag}(y),
$$

$$
\langle x\odot y,z\rangle
=\langle x,y\odot z\rangle
=\langle y,x\odot z\rangle,
$$

et

$$
|x\odot y|\leq |x|\,|y|.
$$

### Vérification avec JAX

Dans JAX, l'opérateur `*` est le produit de Hadamard pour des tableaux de même forme, `jnp.diag(x)` construit $\operatorname{Diag}(x)$ et `jnp.linalg.norm(x)` calcule la norme euclidienne.

Construisez trois vecteurs aléatoires et vérifiez les quatre résultats. Pour les égalités entre nombres flottants, utilisez `jnp.allclose` plutôt que `==`.

## Bilan

Le produit `*` entre tableaux de même forme représente le produit de Hadamard. Pour une fonction appliquée composante par composante, la matrice jacobienne est diagonale et les dérivées directionnelles se calculent elles aussi composante par composante.

La transformation `jax.jvp` donne directement l'action d'une dérivée sur une direction. Elle sera particulièrement utile pour vérifier l'action d'une Hessienne sans former explicitement une grande matrice.
