## 1. Une machine $P^1$ écrite avec ReLU

Une fonction affine par morceaux sur la droite possède une représentation
$$
v(x)=\alpha+\beta x+\sum_{i=1}^N c_i(x-t_i)_+.
$$

Les nœuds $t_i$ sont les points où la pente peut changer ; le coefficient $c_i$ est précisément le saut de pente en $t_i$. La fonction suivante est la version minimale de la machine construite dans le notebook 03.

```python
def machine_p1_relu(x, alpha, beta, noeuds, sauts_pente):
    activations = jax.nn.relu(x[..., None] - noeuds)
    return alpha + beta * x + activations @ sauts_pente
```

## Bilan

La fonction tente est une machine $P^1$, donc un petit réseau ReLU. Sa composition avec elle-même reste facile à évaluer, mais sa description comme fonction affine par morceaux devient rapidement complexe.

Pour représenter $\tau^{\circ L}$, on peut choisir une largeur constante égale à $3$ et une profondeur $L$, ou revenir à une seule couche cachée et utiliser une largeur de l'ordre de $2^L$. Dans cet exemple, la profondeur remplace donc une croissance exponentielle de la largeur.
