## 5. Stabilité numérique

La formule directe peut déborder avant même l'application du logarithme. Comparer
$$
\frac1\beta\log\left(\sum_j e^{\beta x_j}\right)
$$
à la formule obtenue après factorisation de $e^{\beta m}$, avec $m=\max_jx_j$.

```python
def log_somme_exponentielle_naive(x, beta=1.0):
    return jnp.log(jnp.sum(jnp.exp(beta * x))) / beta

x_grand = jnp.array([999.0, 1000.0, 1001.0])
```

## Bilan

Le softmax est le gradient d'une approximation convexe et régulière du maximum. Sa Hessienne est une matrice de covariance pondérée : elle est positive et annule la direction constante. Enfin, soustraire le maximum avant de calculer les exponentielles ne change pas la fonction, mais rend son évaluation beaucoup plus robuste.
