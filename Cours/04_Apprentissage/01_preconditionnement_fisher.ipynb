## 2. Gradient, Hessian and two Fisher matrices

For

\[
s_i(p)=\operatorname{sigmoid}(p^\top x_i),
\qquad
\mathcal L(p)=\frac1n\sum_i
\bigl[\log(1+e^{p^\top x_i})-z_i p^\top x_i\bigr],
\]

compute the gradient and Hessian. Then verify that the Fisher matrix of the Bernoulli model coincides with the Hessian:

\[
F(p)=\frac1n\sum_i s_i(p)(1-s_i(p))x_ix_i^\top.
\]

The empirical Fisher replaces the expectation under the model with the observed labels, containing $(z_i-s_i(p))^2$ instead of $s_i(p)(1-s_i(p))$.

## Summary

- The gradient depends on the chosen metric on parameter space.
- The model Fisher matrix is a metric; the empirical Fisher is a different matrix and generally does not equal it.
- In logistic regression, the Fisher matrix and Hessian coincide.
- Preconditioning acts on conditioning and reduces artificial dependence on the coordinate choice.
