## 2. Gradient, Hessienne et deux matrices de Fisher

Pour

\[
s_i(p)=\operatorname{sigmoid}(p^\top x_i),
\qquad
\mathcal L(p)=\frac1n\sum_i
\bigl[\log(1+e^{p^\top x_i})-z_i p^\top x_i\bigr],
\]

calculez le gradient et la Hessienne. Vérifiez ensuite que la Fisher du modèle
de Bernoulli coïncide avec la Hessienne :

\[
F(p)=\frac1n\sum_i s_i(p)(1-s_i(p))x_ix_i^\top.
\]

La Fisher empirique remplace l'espérance sous le modèle par les étiquettes
observées et contient $(z_i-s_i(p))^2$ à la place de $s_i(p)(1-s_i(p))$.

## Bilan

- Le gradient dépend de la métrique choisie sur l'espace des paramètres.
- La matrice de Fisher du modèle est une métrique ; la Fisher empirique est
  une autre matrice et ne lui est pas égale en général.
- Dans la régression logistique, Fisher et Hessienne coïncident.
- Le préconditionnement agit sur le conditionnement et réduit la dépendance
  artificielle au choix des coordonnées.
