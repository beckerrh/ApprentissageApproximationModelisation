## 2. Two functionals

The energy functional is
\[
J(u)=\frac12\int_0^1 |u'|^2-\int_0^1fu,
\]
while the strong-residual loss is
\[
\mathcal L_{\rm fort}(p)=\frac1N\sum_{i=1}^N
\bigl(-u_p''(x_i)-f(x_i)\bigr)^2.
\]
