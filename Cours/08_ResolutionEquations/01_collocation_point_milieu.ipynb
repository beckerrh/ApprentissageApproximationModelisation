## 1. Linear equation

On $[t_i,t_{i+1}]$, the affine machine satisfies
\[
u_h'(t_{i+\frac12})=\frac{u_{i+1}-u_i}{h},
\qquad
u_h(t_{i+\frac12})=\frac{u_i+u_{i+1}}2.
\]
Imposing the ODE at the midpoint, show that
\[
\left(I-\frac h2A\right)u_{i+1}
=\left(I+\frac h2A\right)u_i+hb.
\]
For constant $b$, this is the Crank--Nicolson scheme, also called the implicit midpoint scheme here.
