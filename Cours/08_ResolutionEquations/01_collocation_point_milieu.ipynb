## 1. Équation linéaire

Sur $[t_i,t_{i+1}]$, la machine affine vérifie
\[
u_h'(t_{i+\frac12})=\frac{u_{i+1}-u_i}{h},
\qquad
u_h(t_{i+\frac12})=\frac{u_i+u_{i+1}}2.
\]
En imposant l'EDO au point milieu, montrer que
\[
\left(I-\frac h2A\right)u_{i+1}
=\left(I+\frac h2A\right)u_i+hb.
\]
Pour $b$ constant, il s'agit du schéma de Crank--Nicolson, également appelé ici schéma du point milieu implicite.
