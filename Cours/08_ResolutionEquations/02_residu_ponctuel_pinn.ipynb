## 1. Une perturbation invisible aux points de collocation

Soient $t_1,\ldots,t_N$ les points de collocation et
\[
w(t)=t\prod_{i=1}^N(t-t_i)^2.
\]
Alors $w(0)=0$ et $w(t_i)=w'(t_i)=0$. Par conséquent,
\[
v_A(t)=e^{-\mu t}+A w(t)
\]
satisfait exactement la condition initiale et possède un résidu nul à tous les points $t_i$, quelle que soit l'amplitude $A$.
