$$
g_p(x)=W_2^{\mathsf T}\tanh(W_1^{\mathsf T}x+b_1)+b_2,

## 4. Margin and robustness

Transform $g_p$ into a binary classifier: the class is determined by the sign of $g_p(x)$. The margin to the threshold is $|g_p(x)|$.

The quantity

$$
r_{\rm lin}=
\frac{|g_p(x)|}{\|\nabla_x g_p(x)\|}
$$

is an estimate from linearisation, but not necessarily a guarantee. Since $|\tanh'|\leq1$, a global Lipschitz constant is

$$
K=\|W_1\|_2\,\|W_2\|_2,

## Summary

The derivatives $D_x\Phi$ and $D_p\Phi$ give local sensitivities. Their norms provide conditioning, which must be distinguished from procedural stability and decision robustness. Shapley values answer another question: they distribute a variation among features according to a precise replacement rule. None of these quantities is independent of the norms, units, reference value or output studied.
