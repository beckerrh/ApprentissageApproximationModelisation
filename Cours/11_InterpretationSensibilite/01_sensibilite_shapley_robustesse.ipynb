## 4. Marge et robustesse

On transforme $g_p$ en classificateur binaire : la classe est déterminée par le signe de $g_p(x)$. La marge au seuil vaut $|g_p(x)|$.

La quantité

$$
r_{\rm lin}=
\frac{|g_p(x)|}{\|\nabla_x g_p(x)\|}
$$

est une estimation issue de la linéarisation, mais pas nécessairement une garantie. Comme $|\tanh'|\leq1$, une constante de Lipschitz globale est

$$
K=\|W_1\|_2\,\|W_2\|_2,
$$

et toute perturbation $h$ telle que $\|h\|<|g_p(x)|/K$ conserve certainement la classe.

### Conditionnement et stabilité

Le conditionnement décrit ici la sensibilité de l'application $g_p$. Si $g_p$ approchait une application de référence $S$, il faudrait comparer $\|D_xg_p(x)\|$ et $\|DS(x)\|$. Une forte sensibilité déjà présente dans $S$ traduit un problème mal conditionné ; une sensibilité supplémentaire introduite par $g_p$ signale une instabilité de la machine.

## Bilan

Les dérivées $D_x\Phi$ et $D_p\Phi$ donnent des sensibilités locales. Leur norme fournit un conditionnement, qui doit être distingué de la stabilité de la procédure et de la robustesse d'une décision. Les valeurs de Shapley répondent à une autre question : elles répartissent une variation entre les variables relativement à une règle précise de remplacement. Aucune de ces quantités n'est indépendante du choix des normes, des unités, de la valeur de référence ou de la sortie étudiée.
