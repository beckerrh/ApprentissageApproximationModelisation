## Bilan

Un hyperplan ne suffit pas à définir un problème d'apprentissage. Il faut encore choisir une normalisation, une perte et une régularisation. Sur des données séparables, la perte logistique non régularisée illustre ce point de façon particulièrement nette : la frontière peut rester fixe tandis que la norme du score diverge.
