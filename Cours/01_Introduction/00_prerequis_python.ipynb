## 2. Listes, tuples, indices et boucles

Un objet est **modifiable** (`'mutable'`) si son contenu peut changer sans que l'on construise un nouvel objet. Les `list` et les `dict` sont modifiables ; les nombres, les chaînes de caractères et les tuples ne le sont pas. Une affectation ne copie pas nécessairement l'objet : deux noms peuvent donc désigner la même liste. Les tuples sont notamment utiles pour regrouper des objets dont le nombre et le rôle sont fixés.

````python
valeurs = [1, 2, 3, 4, 5]
point = (2.0, -1.0)

print(valeurs[0], valeurs[-1], valeurs[1:4])
print(point[0], point[1])

carres_pairs = [k**2 for k in valeurs if k % 2 == 0]
print(carres_pairs)

valeurs.append(6)
dernier = valeurs.pop()
print(f"{dernier=} {valeurs=}")

for indice, valeur in enumerate(valeurs[:3]):
    print(f"{indice=} {valeur=}")

somme = 0
for valeur in valeurs:
    somme += valeur
print(somme)

u = [1, 2]
v = u              # u et v désignent la même liste
w = u.copy()       # w désigne une nouvelle liste
v.append(3)
print(f"{u=} {v=} {w=}")
````

## 3. Dictionnaires

Un dictionnaire associe des clés à des valeurs. Nous en utiliserons souvent pour organiser des paramètres ou des résultats.

````python
parametres = {
    "largeur": 16,
    "profondeur": 3,
    "activation": "tanh",
}

print(parametres["largeur"])

for nom, valeur in parametres.items():
    print(f"{nom}: {valeur}")

parametres_modifies = {**parametres, "largeur": 32}
print(parametres)
print(parametres_modifies)
````

## 4. Fonctions

Il faut savoir définir une fonction, lui transmettre des arguments et utiliser sa valeur de retour.

````python
def erreur_quadratique(prediction, cible):
    if len(prediction) != len(cible):
        raise ValueError("prediction et cible doivent avoir la même longueur")
    erreurs = [(p - z)**2 for p, z in zip(prediction, cible)]
    return sum(erreurs) / len(erreurs)


prediction = [1.0, 2.0, 4.0]
cible = [1.5, 2.0, 3.0]
print(erreur_quadratique(prediction, cible))
````

## 5. Une classe très simple

Une classe regroupe des données et les opérations qui leur sont associées. Une connaissance élémentaire de 'class', '__init__', 'self' et des attributs suffit pour commencer le cours.

````python
class MachineAffine:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def __call__(self, x):
        return self.a * x + self.b

    def nombre_parametres(self):
        return 2


machine = MachineAffine(a=2.0, b=-1.0)
print(machine(3.0))
print(machine.a, machine.b)
print(machine.nombre_parametres())
````

## 6. Listes Python et tableaux numériques

Une liste organise des objets Python. Un tableau numérique possède notamment une forme, un type numérique et des opérations composante par composante. JAX reprendra une grande partie de cette syntaxe.

````python
import numpy as np

liste = [1.0, 2.0, 3.0]
x = np.array(liste)
y = np.array([2.0, -1.0, 4.0])
A = np.array([[1.0, 2.0, 0.0], [0.0, -1.0, 3.0]])

print(liste + liste)       # concaténation de listes
print(x + x)               # addition de tableaux
print(x * y)               # produit de Hadamard
print(A @ x)               # produit matrice-vecteur
print(A.T)                 # transposée
print(x.shape, x.dtype)
print(A.shape, A.dtype)
````
