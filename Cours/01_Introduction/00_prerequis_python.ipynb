## 2. Lists, tuples, indices and loops

An object is **mutable** if its contents can change without creating a new object. Lists and dictionaries are mutable; numbers, strings and tuples are not. An assignment does not necessarily copy the object: two names may therefore refer to the same list. Tuples are particularly useful for grouping objects whose number and roles are fixed.

````python
valeurs = [1, 2, 3, 4, 5]
point = (2.0, -1.0)

print(valeurs[0], valeurs[-1], valeurs[1:4])
print(point[0], point[1])

carres_pairs = [k**2 for k in valeurs if k % 2 == 0]
print(carres_pairs)

valeurs.append(6)
dernier = valeurs.pop()
print(f"{dernier=} {valeurs=}")

for indice, valeur in enumerate(valeurs[:3]):
    print(f"{indice=} {valeur=}")

somme = 0
for valeur in valeurs:
    somme += valeur
print(somme)

u = [1, 2]
v = u              # u and v refer to the same list
w = u.copy()       # w refers to a new list
v.append(3)
print(f"{u=} {v=} {w=}")
````

**Check.** Rewrite the comprehension defining 'carres_pairs' using a loop. Why does 'valeurs[1:4]' not contain 'valeurs[4]'? What do 'append', 'pop' and 'enumerate' do? Finally, explain why a change made through `v` is also visible in `u`, but not in `w`.

## 3. Dictionaries

A dictionary associates keys with values. We shall often use dictionaries to organise parameters or results.

````python
parametres = {
    "width": 16,
    "depth": 3,
    "activation": "tanh",
}

print(parametres["width"])

for nom, valeur in parametres.items():
    print(f"{nom}: {valeur}")

parametres_modifies = {**parametres, "width": 32}
print(parametres)
print(parametres_modifies)
````

## 4. Functions

You should know how to define a function, pass arguments to it and use its return value.

````python
def erreur_quadratique(prediction, cible):
    if len(prediction) != len(cible):
        raise ValueError("prediction and target must have the same length")
    erreurs = [(p - z)**2 for p, z in zip(prediction, cible)]
    return sum(erreurs) / len(erreurs)


prediction = [1.0, 2.0, 4.0]
cible = [1.5, 2.0, 3.0]
print(erreur_quadratique(prediction, cible))
````

**Check.** What does 'zip' do? Modify the function to compute the mean absolute error.

## 5. A very simple class

A class groups data and the operations associated with them. An elementary understanding of 'class', '__init__', 'self' and attributes is sufficient to begin the course.

````python
class MachineAffine:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def __call__(self, x):
        return self.a * x + self.b

    def nombre_parametres(self):
        return 2


machine = MachineAffine(a=2.0, b=-1.0)
print(machine(3.0))
print(machine.a, machine.b)
print(machine.nombre_parametres())
````

**Check.** What is the role of 'self'? Add a method that changes the intercept 'b'.

## 6. Python lists and numerical arrays

A list organises Python objects. A numerical array has, in particular, a shape, a numerical type and componentwise operations. JAX uses much of this syntax.

````python
import numpy as np

liste = [1.0, 2.0, 3.0]
x = np.array(liste)
y = np.array([2.0, -1.0, 4.0])
A = np.array([[1.0, 2.0, 0.0], [0.0, -1.0, 3.0]])

print(liste + liste)       # list concatenation
print(x + x)               # array addition
print(x * y)               # Hadamard product
print(A @ x)               # matrix-vector product
print(A.T)                 # transpose
print(x.shape, x.dtype)
print(A.shape, A.dtype)
````
